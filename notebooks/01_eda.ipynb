# 01. Exploratory Data Analysis (EDA)
## Reddit Financial Sentiment Analysis & Knowledge Distillation
**Reference:** Deng et al. (WWW '23 Companion, arXiv:2212.11311)

This notebook explores the raw and cleaned Reddit financial submissions:
- Dataset distribution across subreddits (`wallstreetbets`, `stocks`, `investing`, `StockMarket`)
- Ticker detection and entity frequency
- Text length distributions and vocabulary
- Sentiment label distribution and temporal coverage

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Ensure project root in sys.path
sys.path.insert(0, str(Path.cwd().parent))

from src.utils.common import DATA_PROC, DATA_RAW
```

```python
# Load processed Reddit dataset
reddit_df = pd.read_parquet(DATA_PROC / "processed_reddit.parquet")
print(f"Loaded {len(reddit_df)} processed Reddit posts.")
reddit_df.head()
```

```python
# Class distribution
print("Sentiment Distribution:")
print(reddit_df["sentiment_label"].value_counts(normalize=True))

# Ticker distribution
print("\nTop 10 Mentioned Tickers:")
print(reddit_df["ticker"].value_counts().head(10))
```

```python
# Plot post length distribution
reddit_df["char_len"] = reddit_df["text"].str.len()
plt.figure(figsize=(8, 4), dpi=150)
plt.hist(reddit_df["char_len"], bins=50, color="#38bdf8", edgecolor="black")
plt.title("Post Character Length Distribution")
plt.xlabel("Characters")
plt.ylabel("Count")
plt.grid(alpha=0.3)
plt.show()
```
